# Rapport – Projet Databricks Alpenia

Une table managée est entièrement gérée par Databricks, tandis qu'une table externe conserve ses données dans un emplacement de stockage défini. Je choisis une table managée pour les données propres au projet et une table externe lorsque les fichiers doivent rester accessibles à d'autres outils.

Les fichiers du volume `landing` sont les données sources dans leur format d'origine (CSV, JSON ou Parquet). Les tables `brz_` sont leurs versions chargées dans Databricks au format Delta.

Delta Lake offre des fonctionnalités intéressantes par rapport à de simples fichiers Parquet. J'ai notamment utilisé `DESCRIBE HISTORY` pour consulter les modifications et `RESTORE` pour revenir à une version précédente.

J'ai choisi le mode `overwrite` pour remplacer les tables à chaque ingestion et garantir un résultat reproductible. Si de nouveaux fichiers arrivaient chaque jour, j'utiliserais plutôt une ingestion incrémentale (`append`), avec un contrôle des doublons.

`DESCRIBE HISTORY` permet de suivre les opérations effectuées sur une table et de retrouver une version antérieure. Dans mon projet, je l'ai utilisé pour examiner les écritures et vérifier une restauration.

Enfin, les notebooks utilisent le compute serverless associé aux notebooks, tandis que l'éditeur SQL utilise un SQL Warehouse serverless dans mon workspace Databricks Free Edition.
